```python
#creating a DNA sequence
dna = "ATGCGATACGTA"
print(dna)
```

```python
#sequence type
type(dna)
```

```python
#sequence length
print(len(dna))
```

```python
#counting Nuleotides
print("A:", dna.count("A"))
print("T:", dna.count("T"))
print("G:", dna.count("G"))
print("C:", dna.count("C"))
```

```python
#visualization
import matplotlib.pyplot as plt
A = dna.count("A")
T = dna.count("T")
G = dna.count("G")
C = dna.count("C")

nuleotides =["A","T","G","C"]
counts = [A,T,G,C]
plt.bar(nuleotides,counts)
plt.xlabel("Nuleotide")
plt.ylabel("count")
plt.title("DNA Nuleotide Composition")

plt.show()
```

```python
#GC Content
#GC Content=((G+C)/Total_Nuleotides)x100

gc_count=dna.count("G")+dna.count("C")
gc_content=(gc_count/len(dna))*100
print("GC Content:",gc_content,"%")
```

```python
#DNA -> RNA
rna=dna.replace("T","U")
print("DNA:",dna)
print("RNA:",rna)
```

```python
#translate - applies that mapping to the entire sequence
complement=dna.translate(str.maketrans("ATGC","TACG"))
print(complement)
```

```python
#Reverse Complement
reverse_complement=complement[::-1]
#take the sequence from begining to end with a step of -1
print(reverse_complement)
```
