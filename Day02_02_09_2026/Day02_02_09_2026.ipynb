```python
#find the numbers of patients
len(df)
```

```python
#How many patients have each tumor type
df["TumorType"].value_counts()
```

```python
#find the average survival for each tumor type
df.groupby("TumorType")["SurvivalInMonths"].mean()
```

```python
#count between tumor types
df.groupby("TumorType")["MutationCount"].mean()
```

```python
#count patients receiving each immunotherapy
df["ImmunotherapyType"].value_counts()
```

```python
#compare survial between immunotherapy types
df.groupby("ImmunotherapyType")["SurvivalInMonths"].mean()
```

```python
#compare treatment response by immunotherapy
pd.crosstab(
    df["ImmunotherapyType"],
    df["TreatmentResponse"]
)
```

```python
#count patients by biomarker expression
df["BiomarkerExpressionLevel"].value_counts()
```

```python
#compare survial according to biomarker level
df.groupby("BiomarkerExpressionLevel")["SurvivalInMonths"].mean()
```

```python
#compare Biomarker Expression withtreatment response
pd.crosstab(
    df["BiomarkerExpressionLevel"],
    df["TreatmentResponse"]
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#how many patients belong to each cancer type? -  Bar chart
sns.countplot(data=df,x="TumorType")
plt.title("Number of patients by Tumor Type")
plt.xlabel("Tumor Type")
plt.ylabel("Number of Patients")
plt.show()
```

```python
#Box plot - suvial by tumor type
sns.boxplot(
    data=df,
    x="TumorType",
    y="SurvivalInMonths"
)
plt.title("Survial by Tumor Type")
plt.show()
```

```python
#Scatter plot - Mutation count vs survial
sns.scatterplot(
    data=df,
    x="MutationCount",
    y="SurvivalInMonths",
    hue="TreatmentResponse"
)
plt.title("Mutation count vs survial")
plt.xlabel("Mutation count")
plt.ylabel("survial (Months)")
plt.show()
```

```python
#Heatmap
numeric_df=df[
    ["Age","SurvivalInMonths","MutationCount"]
]
correlation=numeric_df.corr()
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```
