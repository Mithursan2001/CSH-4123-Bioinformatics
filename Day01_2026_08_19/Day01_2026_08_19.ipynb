```python
sequence = "ATGCGATCGX"
if "X" in sequence:
    print("Invalid DNA sequence")
else:
    print("Valid DNA sequence")
```

```python
sequence = "ATGCGATCG"
valid_bases = "ATGC"
valid = True

for base in sequence:
    if base not in valid_bases:
        valid = False
if valid:
    print("valid DNA sequence")
else:
    print("Invalid DNA sequence")
```

```python
import pandas as pd
data={
    "Gene":["BRCA1","TP53","EGFR","KRAS"],
    "Organism":["Human","Human","Human","Human"],
    "Length":[1200,1500,1100,900],
    "GC_content":[45.2,51.3,48.7,42.1]
}
df=pd.DataFrame(data)
print(df)
```

```python
#filtering the biological data
df[df["GC_content"]>45]
```

```python
#sorting
df.sort_values("GC_content")
df.sort_values("GC_content",ascending=False)
```

```python
#add new column
df["Length_kb"]=df["Length"]/100

print(df)
```

```python
#Find gens with length greater than 1100
df[df["Length"]>1100]
```

```python
#count genes with length greater than 1000
(df["Length"]>1000).sum()
```

```python
#find genes with both length > 1000 and GC content > 45
df[(df["Length"]>1000) & (df["GC_content"]>45)]
```

```python
#compare average GC content by organism
df.groupby("Organism")["GC_content"].mean()
```
