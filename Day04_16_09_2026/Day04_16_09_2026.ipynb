```python
from Bio.Seq import Seq
dna = Seq("ATGCGATCGATCGAT")
print(dna)
```

```python
print(type(sequence))
print(type(dna))
```

```python
from Bio.Seq import Seq

print("Sequence ",dna)
print("Length ",len(dna))
print("First ",dna[0])
print("Last",dna[-1])
```

```python
#Transcription
rna=dna.transcribe()
print("DNA:",dna)
print("RNA:",rna)
```

```python
#reverse transcription - converts an RNA sequence to
back_to_dna = rna.back_transcribe()
print(back_to_dna)
```

```python
#translation
rna=dna.transcribe()
protein=dna.translate()
print("DNA:",dna)
print("RNA:",rna)
print("Protein: ",protein)
```

```python
# Retrive the BRCA1 sequence from NCBI
from Bio import SeqIO

record = SeqIO.read("sequence.fasta","fasta")
print("ID:",record.id)
print("Description:",record.description)
print("Sequence:",record.seq)
print("Length:",len(record.seq))
```

```python
print("Length: ",len(record.seq))
print("A:", record.seq.count("A"))
print("T:", record.seq.count("T"))
print("G:", record.seq.count("G"))
print("C:", record.seq.count("C"))
gc_count=record.seq.count("G")+record.seq.count("C")
gc_content=(gc_count/len(record.seq))*100
print("GC Content:",gc_content,"%")
```

```python
from Bio import SeqIO

rna=record.seq.transcribe()
print("DNA:\n",record.seq)
print("RNA:\n",rna)
```

```python
back_to_dna = record.seq.back_transcribe()
print(back_to_dna)
```

```python
#translation
rna=record.seq.transcribe()
protein=record.seq.translate()
print("DNA:\n",record.seq)
print("RNA:\n",rna)
print("Protein: \n",protein)
```
